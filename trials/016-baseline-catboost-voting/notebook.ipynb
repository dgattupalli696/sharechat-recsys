# T016 - Original CatBoost validation baseline in soft voting

No retraining, test inference or submission. All primary holdout rows. Membership/weights are holdout-selected.

```
from pathlib import Path
import subprocess
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'pyproject.toml').is_file() and (p / 'data/train.parquet').is_file())
TRIAL = ROOT / 'trials/016-baseline-catboost-voting'
OUTPUT = ROOT / 'artifacts/local-t016'
subprocess.run([
    sys.executable, str(TRIAL / 'frozen/predict_catboost_validation.py'),
    '--data', str(ROOT / 'data/train.parquet'),
    '--model', str(ROOT / 'artifacts/kaggle-v2/validation.cbm'),
    '--identities', str(ROOT / 'artifacts/kaggle-t014/validation-identities.parquet'),
    '--output-dir', str(ROOT / 'artifacts/b001-validation-replay'),
    '--expected-auc', '0.8160260803',
], cwd=ROOT, check=True)
subprocess.run([sys.executable, str(TRIAL / 'prepare_inputs.py')], cwd=ROOT, check=True)
subprocess.run([
    sys.executable, str(TRIAL / 'frozen/ensemble_experiment.py'),
    '--manifest', str(OUTPUT / 'candidates.json'),
    '--identities', str(ROOT / 'artifacts/kaggle-t014/validation-identities.parquet'),
    '--output-dir', str(OUTPUT),
    '--previous-predictions', str(ROOT / 'artifacts/local-t015/validation_predictions.npz'),
], cwd=ROOT, check=True)
```
