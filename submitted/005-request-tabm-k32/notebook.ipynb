```python
import argparse
import gc
import json
from pathlib import Path
import time

from catboost import CatBoostClassifier
from catboost.utils import get_gpu_device_count
import numpy as np
import polars as pl
import pyarrow.parquet as pq
from sklearn.metrics import roc_auc_score


ROOT = Path(__file__).resolve().parent
ON_KAGGLE = Path("/kaggle/input").is_dir()
TARGET = "watched"
TIME = "feature1"


def features(frame, expand_embeddings=False):
    expressions = []
    for name, dtype in frame.schema.items():
        if name in {TARGET, "request_id", "id"}:
            continue
        column = pl.col(name)
        if isinstance(dtype, pl.Datetime):
            expressions.extend(
                [
                    column.dt.hour().cast(pl.Float32).alias(f"{name}_hour"),
                    column.dt.weekday().cast(pl.Float32).alias(f"{name}_weekday"),
                ]
            )
        elif isinstance(dtype, pl.List):
            expressions.extend(
                [
                    column.list.len().cast(pl.Float32).alias(f"{name}_length"),
                    column.list.mean().cast(pl.Float32).alias(f"{name}_mean"),
                    column.list.std().cast(pl.Float32).alias(f"{name}_std"),
                    column.list.min().cast(pl.Float32).alias(f"{name}_min"),
                    column.list.max().cast(pl.Float32).alias(f"{name}_max"),
                ]
            )
            if expand_embeddings:
                invalid = frame.select(
                    (column.is_not_null() & (column.list.len() != 32)).any()
                ).item()
                if invalid:
                    raise ValueError(f"{name} must contain 32-element vectors")
                expressions.extend(
                    column.list.get(index).cast(pl.Float32).alias(f"{name}_dim_{index}")
                    for index in range(32)
                )
        elif dtype == pl.String:
            expressions.append(column.fill_null("__MISSING__").alias(name))
        elif dtype.is_numeric():
            expressions.append(column.cast(pl.Float32).alias(name))
        else:
            raise TypeError(f"Unsupported feature {name}: {dtype}")
    return frame.select(expressions).with_columns(
        pl.col(pl.Float32).fill_nan(None)
    )


def load_sample(path, maximum_rows, seed, expand_embeddings=False):
    parquet = pq.ParquetFile(path)
    total = parquet.metadata.num_rows
    rng = np.random.default_rng(seed)
    indices = np.sort(rng.choice(total, min(maximum_rows, total), replace=False))
    frames = []
    offset = 0
    for batch in parquet.iter_batches(batch_size=8192):
        end = offset + batch.num_rows
        selected = indices[(indices >= offset) & (indices < end)] - offset
        if len(selected):
            raw = pl.from_arrow(batch.take(selected))
            frames.append(
                features(raw, expand_embeddings).with_columns(
                    raw[TARGET], raw[TIME].alias("_time")
                )
            )
        offset = end
    # Callers sort or slice next; skipping the rechunk avoids a full extra copy.
    return pl.concat(frames, rechunk=False)


def model(iterations, threads, task_type):
    return CatBoostClassifier(
        iterations=iterations,
        depth=6,
        learning_rate=0.06,
        loss_function="Logloss",
        eval_metric="AUC",
        random_seed=42,
        thread_count=threads,
        task_type=task_type,
        max_ctr_complexity=1,
        allow_writing_files=False,
        verbose=50,
    )


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--rows", type=int, default=1_600_000 if ON_KAGGLE else 150_000)
    parser.add_argument("--iterations", type=int, default=400)
    parser.add_argument("--threads", type=int, default=4)
    parser.add_argument("--expand-embeddings", action="store_true")
    parser.add_argument("--validation-only", action="store_true")
    parser.add_argument(
        "--task-type", choices=["CPU", "GPU"], default="GPU" if ON_KAGGLE else "CPU"
    )
    parser.add_argument("--data-dir", type=Path)
    parser.add_argument(
        "--output-dir",
        type=Path,
        default=Path("/kaggle/working") if ON_KAGGLE else ROOT / "artifacts",
    )
    args = parser.parse_args()
    if min(args.rows, args.iterations, args.threads) <= 0:
        parser.error("rows, iterations, and threads must be positive")

    if args.task_type == "GPU" and get_gpu_device_count() == 0:
        raise RuntimeError("GPU training requested, but no CUDA GPU is available")
    data = args.data_dir
    if data is None and ON_KAGGLE:
        candidates = [
            path.parent
            for path in Path("/kaggle/input").rglob("train.parquet")
            if (path.parent / "test.parquet").is_file()
            and (path.parent / "sample_submission.csv").is_file()
        ]
        if len(candidates) != 1:
            raise RuntimeError(f"Expected one ShareChat dataset, found {candidates}")
        data = candidates[0]
    elif data is None:
        data = ROOT / "data"
    output = args.output_dir
    output.mkdir(parents=True, exist_ok=True)
    print(f"Compute: {args.task_type}; input: {data}; output: {output}", flush=True)
    sample = load_sample(
        data / "train.parquet", args.rows, 42, args.expand_embeddings
    ).sort("_time")
    if sample.height < 2:
        raise ValueError("Chronological validation requires at least two rows")
    if sample["_time"].null_count():
        raise ValueError("Missing timestamps prevent chronological validation")
    cutoff = sample["_time"][int(sample.height * 0.8)]
    train = sample.filter(pl.col("_time") < cutoff)
    valid = sample.filter(pl.col("_time") >= cutoff)
    if train.is_empty() or valid.is_empty():
        raise ValueError(
            f"Timestamp cutoff {cutoff} cannot produce nonempty chronological splits"
        )
    for name, part in [("train", train), ("validation", valid)]:
        if part[TARGET].n_unique() != 2:
            raise ValueError(f"{name} must contain both target classes")
    columns = [name for name in sample.columns if name not in {TARGET, "_time"}]
    categories = [name for name in columns if sample.schema[name] == pl.String]
    x_train = train.select(columns).to_pandas()
    x_valid = valid.select(columns).to_pandas()
    y_train = train[TARGET].to_numpy()
    y_valid = valid[TARGET].to_numpy()
    sample_rows, training_rows, validation_rows = sample.height, train.height, valid.height
    print(
        f"Training: {train.height:,}; validation: {valid.height:,}; "
        f"features: {len(columns)}; cutoff: {cutoff}",
        flush=True,
    )
    if args.validation_only:
        del sample, train, valid
        gc.collect()
    estimator = model(args.iterations, args.threads, args.task_type)
    fit_started = time.monotonic()
    estimator.fit(
        x_train,
        y_train,
        cat_features=categories,
        eval_set=(x_valid, y_valid),
        early_stopping_rounds=60,
    )
    fit_seconds = time.monotonic() - fit_started
    auc = roc_auc_score(
        y_valid, estimator.predict_proba(x_valid)[:, 1]
    )
    best_iterations = estimator.tree_count_
    metrics = {
        "validation_auc": auc,
        "validation": "chronological 80/20 split; timestamp ties kept together",
        "cutoff": str(cutoff),
        "sample_rows": sample_rows,
        "training_rows": training_rows,
        "validation_rows": validation_rows,
        "iterations": best_iterations,
        "max_iterations": args.iterations,
        "expand_embeddings": args.expand_embeddings,
        "validation_only": args.validation_only,
        "fit_seconds": fit_seconds,
        "task_type": args.task_type,
        "features": columns,
        "categorical_features": categories,
    }
    estimator.save_model(str(output / "validation.cbm"))
    (output / "metrics.json").write_text(json.dumps(metrics, indent=2) + "\n")
    print(f"Validation ROC AUC: {auc:.6f}", flush=True)
    if args.validation_only:
        print("Validation-only run: skipping full-data refit and test predictions", flush=True)
        return
    del estimator, x_train, x_valid, train, valid
    gc.collect()

    print("Refitting on the full sample", flush=True)
    estimator = model(best_iterations, args.threads, args.task_type)
    estimator.fit(
        sample.select(columns).to_pandas(),
        sample[TARGET].to_numpy(),
        cat_features=categories,
    )
    estimator.save_model(str(output / "baseline.cbm"))
    del sample
    gc.collect()

    submission = pl.read_csv(data / "sample_submission.csv")
    if submission.columns != ["id", TARGET]:
        raise ValueError(f"Unexpected submission schema: {submission.columns}")
    test = pq.ParquetFile(data / "test.parquet")
    if test.metadata.num_rows != submission.height:
        raise ValueError("Test and submission row counts differ")
    predictions = []
    offset = 0
    for batch in test.iter_batches(batch_size=8192):
        raw = pl.from_arrow(batch)
        if "id" not in raw.columns:
            raise ValueError("Test data has no id column for submission alignment")
        expected = submission["id"].slice(offset, raw.height)
        if raw["id"].to_list() != expected.to_list():
            raise ValueError("Test ids do not match sample submission order")
        predictions.append(
            estimator.predict_proba(
                features(raw, args.expand_embeddings).select(columns).to_pandas()
            )[:, 1]
        )
        offset += raw.height
    probabilities = np.concatenate(predictions)
    if not np.isfinite(probabilities).all() or not (
        (probabilities >= 0) & (probabilities <= 1)
    ).all():
        raise ValueError("Invalid predicted probabilities")
    submission.with_columns(pl.Series(TARGET, probabilities)).write_csv(
        output / "submission.csv"
    )
    print(f"Saved {submission.height:,} predictions to {output / 'submission.csv'}")

```

```python
import argparse
import gc
import hashlib
from importlib.metadata import version
import json
from pathlib import Path
import time

import joblib
import numpy as np
import polars as pl
from sklearn.impute import SimpleImputer
from sklearn.metrics import log_loss, roc_auc_score
from sklearn.preprocessing import OrdinalEncoder, StandardScaler
import torch
from torch import nn
from torch.nn import functional as F



def numeric_float32(frame, columns, block=64):
    # Column blocks avoid materializing the whole matrix as float64 first.
    array = np.empty((frame.height, len(columns)), dtype=np.float32)
    for start in range(0, len(columns), block):
        names = columns[start : start + block]
        array[:, start : start + len(names)] = frame.select(names).to_numpy()
    return array


def prepare_features(train, valid):
    columns = [name for name in train.columns if name not in {TARGET, "_time"}]
    numeric = [name for name in columns if train.schema[name].is_numeric()]
    categorical = [name for name in columns if train.schema[name] == pl.String]
    if len(numeric) + len(categorical) != len(columns):
        raise TypeError("The engineered sample must contain only numeric/string features")
    imputer = SimpleImputer(
        strategy="median", keep_empty_features=True, add_indicator=True, copy=False
    )
    scaler = StandardScaler(copy=False)
    transformed, infinity_counts = [], []
    for label, frame in [("training", train), ("validation", valid)]:
        array = numeric_float32(frame, numeric)
        count = int(np.isinf(array).sum())
        infinity_counts.append(count)
        if count:
            print(f"Treating {count} infinite {label} values as missing", flush=True)
            array[np.isinf(array)] = np.nan
        if label == "training":
            array = imputer.fit_transform(array)
            array = scaler.fit_transform(array)
        else:
            array = imputer.transform(array)
            array = scaler.transform(array)
        transformed.append(array)
    x_train, x_valid = transformed
    for array in [x_train, x_valid]:
        np.clip(array, -10, 10, out=array)
        if not np.isfinite(array).all():
            raise ValueError("Nonfinite numerical features after preprocessing")
    encoder = OrdinalEncoder(
        handle_unknown="use_encoded_value", unknown_value=-1, dtype=np.int64
    )
    categories = [
        frame.select(categorical).fill_null("__MISSING__").to_numpy()
        for frame in [train, valid]
    ]
    c_train = encoder.fit_transform(categories[0]) + 1
    c_valid = encoder.transform(categories[1]) + 1
    cardinalities = [len(values) + 1 for values in encoder.categories_]
    preprocessing = {
        "numeric_columns": numeric,
        "categorical_columns": categorical,
        "imputer": imputer,
        "scaler": scaler,
        "encoder": encoder,
        "cardinalities": cardinalities,
        "infinity_counts": infinity_counts,
        "validation_unknown_category_rates": {
            name: float((c_valid[:, index] == 0).mean())
            for index, name in enumerate(categorical)
        },
    }
    return (
        torch.from_numpy(np.ascontiguousarray(x_train, dtype=np.float32)),
        torch.from_numpy(np.ascontiguousarray(c_train)),
        torch.from_numpy(np.ascontiguousarray(x_valid, dtype=np.float32)),
        torch.from_numpy(np.ascontiguousarray(c_valid)),
        preprocessing,
    )


def transform_features(frame, state):
    numeric = numeric_float32(frame, state["numeric_columns"])
    infinities = np.isinf(numeric)
    if infinities.any():
        print(f"Treating {int(infinities.sum())} infinite inference values as missing", flush=True)
        numeric[infinities] = np.nan
    numeric = state["scaler"].transform(state["imputer"].transform(numeric))
    np.clip(numeric, -10, 10, out=numeric)
    if not np.isfinite(numeric).all():
        raise ValueError("Nonfinite numerical features after inference preprocessing")
    categories = (
        frame.select(state["categorical_columns"]).fill_null("__MISSING__").to_numpy()
    )
    encoded = state["encoder"].transform(categories) + 1
    return (
        torch.from_numpy(np.ascontiguousarray(numeric, dtype=np.float32)),
        torch.from_numpy(np.ascontiguousarray(encoded, dtype=np.int64)),
    )


class TabMClassifier(nn.Module):
    def __init__(
        self, n_numeric, cardinalities, members=8, width=128, blocks=2, dropout=0.1,
        category_dropout=0.0, positive_rate=None,
        embedding_dimensions=None, numeric_dropout=0.0, dropout_columns=None,
    ):
        super().__init__()
        from tabm import TabM

        if not 0 <= category_dropout <= 1:
            raise ValueError("category_dropout must be between zero and one")
        self.category_dropout = category_dropout
        self.numeric_dropout = numeric_dropout
        self.dropout_columns = dropout_columns or []
        dimensions = embedding_dimensions or [8] * len(cardinalities)
        if len(dimensions) != len(cardinalities) or any(d <= 0 for d in dimensions):
            raise ValueError("Expected one positive embedding dimension per category")
        self.embeddings = nn.ModuleList(
            nn.Embedding(cardinality, dimension, padding_idx=0)
            for cardinality, dimension in zip(cardinalities, dimensions)
        )
        for embedding in self.embeddings:
            nn.init.normal_(embedding.weight, std=0.02)
            with torch.no_grad():
                embedding.weight[0].zero_()
        self.backbone = TabM.make(
            n_num_features=n_numeric + sum(dimensions),
            d_out=1,
            k=members,
            n_blocks=blocks,
            d_block=width,
            dropout=dropout,
        )
        if positive_rate is not None:
            if not 0 < positive_rate < 1:
                raise ValueError("positive_rate must be strictly between zero and one")
            with torch.no_grad():
                self.backbone.output.bias.fill_(
                    float(np.log(positive_rate / (1 - positive_rate)))
                )

    def forward(self, numeric, categorical):
        if self.training and self.numeric_dropout:
            numeric = numeric.clone()
            numeric[:, self.dropout_columns] = F.dropout(
                numeric[:, self.dropout_columns], p=self.numeric_dropout, training=True
            )
        if self.training and self.category_dropout:
            categorical = categorical.masked_fill(
                torch.rand_like(categorical, dtype=torch.float32) < self.category_dropout,
                0,
            )
        embedded = [
            layer(categorical[:, index])
            for index, layer in enumerate(self.embeddings)
        ]
        return self.backbone(torch.cat([numeric, *embedded], dim=1)).squeeze(-1)


@torch.inference_mode()
def predict(model, numeric, categorical, device, batch_size):
    model.eval()
    probabilities = []
    for start in range(0, len(numeric), batch_size):
        logits = model(
            numeric[start : start + batch_size].to(device),
            categorical[start : start + batch_size].to(device),
        )
        probabilities.append(logits.sigmoid().mean(dim=1).cpu().numpy())
    return np.concatenate(probabilities)


def fit_model(
    model, model_config, optimizer, x_train, c_train, y_train, x_valid, c_valid, y_valid,
    *, device, output, epochs, patience, batch_size,
):
    generator = torch.Generator().manual_seed(42)
    best_auc, best_epoch, stale = -np.inf, 0, 0
    history = []
    training_started = time.monotonic()
    for epoch in range(1, epochs + 1):
        epoch_started = time.monotonic()
        model.train()
        order = torch.randperm(len(y_train), generator=generator)
        loss_sum = 0.0
        for indices in order.split(batch_size):
            optimizer.zero_grad(set_to_none=True)
            logits = model(x_train[indices].to(device), c_train[indices].to(device))
            targets = y_train[indices].to(device).unsqueeze(1).expand_as(logits)
            loss = F.binary_cross_entropy_with_logits(logits, targets)
            if not torch.isfinite(loss).item():
                raise ValueError(f"Nonfinite training loss at epoch {epoch}")
            loss.backward()
            optimizer.step()
            loss_sum += loss.item() * len(indices)
        probabilities = predict(model, x_valid, c_valid, device, batch_size)
        auc = float(roc_auc_score(y_valid, probabilities))
        record = {
            "epoch": epoch,
            "training_loss": loss_sum / len(y_train),
            "validation_auc": auc,
            "seconds": time.monotonic() - epoch_started,
        }
        history.append(record)
        print(json.dumps(record), flush=True)
        if auc > best_auc:
            best_auc, best_epoch, stale = auc, epoch, 0
            torch.save(
                {
                    "model_config": model_config,
                    "state_dict": {
                        key: value.detach().cpu().clone()
                        for key, value in model.state_dict().items()
                    },
                },
                output / "best_model.pt",
            )
        else:
            stale += 1
            if stale >= patience:
                break
    training_seconds = time.monotonic() - training_started
    checkpoint = torch.load(output / "best_model.pt", map_location="cpu", weights_only=True)
    model.load_state_dict(checkpoint["state_dict"])
    probabilities = predict(model, x_valid, c_valid, device, batch_size)
    restored_auc = float(roc_auc_score(y_valid, probabilities))
    if abs(restored_auc - best_auc) > 1e-5:
        raise ValueError("Reloaded checkpoint does not reproduce the selected validation AUC")
    np.savez_compressed(
        output / "validation_predictions.npz", y_true=y_valid, probabilities=probabilities
    )
    return {
        "validation_auc": restored_auc,
        "best_epoch": best_epoch,
        "epochs_run": len(history),
        "training_seconds": training_seconds,
        "history": history,
    }


def fit_model_steps(
    model, model_config, optimizer, x_train, c_train, y_train, x_valid, c_valid, y_valid,
    *, device, output, epochs=30, patience=5, batch_size=1024, validation_interval=300,
    seed=42, scheduler=None,
    fixed_budget=False,
):
    if min(epochs, patience, batch_size, validation_interval) <= 0:
        raise ValueError("Training budgets and validation interval must be positive")
    output.mkdir(parents=True, exist_ok=True)
    generator = torch.Generator().manual_seed(seed)
    probe_indices = torch.randperm(
        len(y_train), generator=torch.Generator().manual_seed(42)
    )[: min(16384, len(y_train))]
    steps_per_epoch = (len(y_train) + batch_size - 1) // batch_size
    best_auc, best_step, global_step = -np.inf, 0, 0
    history, checks = [], []
    started = time.monotonic()
    if device == "cuda":
        torch.cuda.reset_peak_memory_stats()
    for epoch in range(1, epochs + 1):
        model.train()
        loss_sum, rows_seen, stop = 0.0, 0, False
        order = torch.randperm(len(y_train), generator=generator)
        for batch_index, indices in enumerate(order.split(batch_size), 1):
            optimizer.zero_grad(set_to_none=True)
            logits = model(x_train[indices].to(device), c_train[indices].to(device))
            targets = y_train[indices].to(device).unsqueeze(1).expand_as(logits)
            loss = F.binary_cross_entropy_with_logits(logits, targets)
            if not torch.isfinite(loss).item():
                raise ValueError(f"Nonfinite loss at epoch {epoch}, step {global_step}")
            loss.backward()
            optimizer.step()
            if scheduler is not None:
                scheduler.step()
            global_step += 1
            loss_sum += loss.item() * len(indices)
            rows_seen += len(indices)
            expired = global_step - best_step >= patience * steps_per_epoch
            if fixed_budget:
                if epoch != epochs or batch_index != steps_per_epoch:
                    continue
            elif (
                global_step % validation_interval != 0
                and batch_index != steps_per_epoch and not expired
            ):
                continue
            probabilities = predict(model, x_valid, c_valid, device, batch_size)
            auc = float(roc_auc_score(y_valid, probabilities))
            probe = predict(
                model, x_train[probe_indices], c_train[probe_indices], device, batch_size
            )
            probe_y = y_train[probe_indices].numpy()
            record = {
                "epoch": epoch,
                "global_step": global_step,
                "fractional_epoch": epoch - 1 + batch_index / steps_per_epoch,
                "validation_auc": auc,
                "validation_logloss": float(log_loss(y_valid, probabilities, labels=[0, 1])),
                "training_probe_auc": float(roc_auc_score(probe_y, probe)),
                "training_probe_logloss": float(log_loss(probe_y, probe, labels=[0, 1])),
                "learning_rate": optimizer.param_groups[0]["lr"],
                "seconds": time.monotonic() - started,
            }
            checks.append(record)
            print(json.dumps(record), flush=True)
            if auc > best_auc:
                best_auc, best_step = auc, global_step
                torch.save(
                    {
                        "model_config": model_config,
                        "state_dict": {
                            key: value.detach().cpu().clone()
                            for key, value in model.state_dict().items()
                        },
                        "selection": record,
                    },
                    output / "best_model.pt",
                )
            model.train()
            if global_step - best_step >= patience * steps_per_epoch:
                stop = True
                break
        history.append({
            "epoch": epoch, "training_loss": loss_sum / rows_seen,
            "last_validation_auc": checks[-1]["validation_auc"] if checks else None,
        })
        if stop:
            break
    checkpoint = torch.load(output / "best_model.pt", map_location="cpu", weights_only=True)
    model.load_state_dict(checkpoint["state_dict"])
    probabilities = predict(model, x_valid, c_valid, device, batch_size)
    auc = float(roc_auc_score(y_valid, probabilities))
    if abs(auc - best_auc) > 1e-5:
        raise ValueError("Fine checkpoint reload does not reproduce its validation AUC")
    np.savez_compressed(
        output / "validation_predictions.npz", y_true=y_valid, probabilities=probabilities
    )
    return {
        "validation_auc": auc,
        "validation_logloss": float(log_loss(y_valid, probabilities, labels=[0, 1])),
        "best_epoch": checkpoint["selection"]["epoch"],
        "best_global_step": best_step,
        "best_fractional_epoch": checkpoint["selection"]["fractional_epoch"],
        "optimizer_steps": global_step, "epochs_run": len(history),
        "training_seconds": time.monotonic() - started,
        "peak_cuda_bytes": torch.cuda.max_memory_allocated() if device == "cuda" else None,
        "history": history, "checkpoint_history": checks,
        "selection_mode": "fixed_budget" if fixed_budget else "validation_auc",
    }


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--rows", type=int, default=150_000)
    parser.add_argument("--sample-path", type=Path)
    parser.add_argument("--epochs", type=int, default=20)
    parser.add_argument("--patience", type=int, default=5)
    parser.add_argument("--batch-size", type=int, default=1024)
    parser.add_argument("--members", type=int, default=8)
    parser.add_argument("--width", type=int, default=128)
    parser.add_argument("--blocks", type=int, default=2)
    parser.add_argument("--threads", type=int, default=4)
    parser.add_argument("--device", choices=["auto", "cpu", "mps", "cuda"], default="auto")
    parser.add_argument("--output-dir", type=Path, default=ROOT / "artifacts/local-tabm-v1")
    args = parser.parse_args()
    if min(
        args.rows, args.epochs, args.patience, args.batch_size,
        args.members, args.width, args.blocks, args.threads,
    ) <= 0:
        parser.error("All numerical arguments must be positive")
    device = args.device
    if device == "auto":
        device = (
            "cuda" if torch.cuda.is_available()
            else "mps" if torch.backends.mps.is_available()
            else "cpu"
        )
    if device == "cuda" and not torch.cuda.is_available():
        raise RuntimeError("CUDA requested but unavailable")
    if device == "mps" and not torch.backends.mps.is_available():
        raise RuntimeError("Apple MPS requested but unavailable")
    torch.set_num_threads(args.threads)
    torch.manual_seed(42)
    started = time.monotonic()
    sample_hash = None
    if args.sample_path:
        sample = pl.read_parquet(args.sample_path)
        if sample.height != args.rows:
            raise ValueError("Cached sample row count does not match --rows")
        with args.sample_path.open("rb") as stream:
            sample_hash = hashlib.file_digest(stream, "sha256").hexdigest()
    else:
        sample = load_sample(ROOT / "data/train.parquet", args.rows, 42)
    sample = sample.select(
        name for name in sample.columns if "_dim_" not in name
    ).sort("_time")
    if sample.height < 2 or sample["_time"].null_count():
        raise ValueError("Chronological validation requires nonmissing timestamps")
    cutoff = sample["_time"][int(sample.height * 0.8)]
    train = sample.filter(pl.col("_time") < cutoff)
    valid = sample.filter(pl.col("_time") >= cutoff)
    for label, frame in [("training", train), ("validation", valid)]:
        if set(frame[TARGET].unique().to_list()) != {0, 1}:
            raise ValueError(f"{label} must contain both binary target classes")
    training_rows, validation_rows = train.height, valid.height
    y_train = torch.tensor(train[TARGET].to_numpy().copy(), dtype=torch.float32)
    y_valid = valid[TARGET].to_numpy().copy()
    x_train, c_train, x_valid, c_valid, preprocessing = prepare_features(train, valid)
    feature_count = sample.width - 2
    sample_rows = sample.height
    del sample, train, valid
    gc.collect()
    output = args.output_dir
    output.mkdir(parents=True, exist_ok=True)
    joblib.dump(preprocessing, output / "preprocessing.joblib")
    model_config = {
        "n_numeric": x_train.shape[1],
        "cardinalities": preprocessing["cardinalities"],
        "members": args.members,
        "width": args.width,
        "blocks": args.blocks,
        "dropout": 0.1,
    }
    model = TabMClassifier(**model_config).to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=0.002, weight_decay=0.0003)
    preparation_seconds = time.monotonic() - started
    print(
        f"Device: {device}; sample: {sample_rows:,}; train: {training_rows:,}; "
        f"valid: {validation_rows:,}; features: {feature_count}; "
        f"cutoff: {cutoff}; preparation: {preparation_seconds:.1f}s",
        flush=True,
    )
    fitted = fit_model(
        model, model_config, optimizer, x_train, c_train, y_train, x_valid, c_valid, y_valid,
        device=device, output=output, epochs=args.epochs, patience=args.patience,
        batch_size=args.batch_size,
    )
    report = {
        "algorithm": "TabM with learned categorical embeddings",
        **fitted,
        "seed": 42,
        "sample_rows": sample_rows,
        "sample_sha256": sample_hash,
        "training_rows": training_rows,
        "validation_rows": validation_rows,
        "cutoff": str(cutoff),
        "input_features": feature_count,
        "numerical_features": len(preprocessing["numeric_columns"]),
        "numerical_features_after_imputation": x_train.shape[1],
        "categorical_features": len(preprocessing["categorical_columns"]),
        "validation_unknown_category_rates": preprocessing["validation_unknown_category_rates"],
        "infinite_input_counts_train_valid": preprocessing["infinity_counts"],
        "model_config": model_config,
        "optimizer": {"name": "AdamW", "learning_rate": 0.002, "weight_decay": 0.0003},
        "max_epochs": args.epochs,
        "patience": args.patience,
        "batch_size": args.batch_size,
        "device": device,
        "threads": args.threads,
        "preparation_seconds": preparation_seconds,
        "versions": {
            name: version(name)
            for name in ["torch", "tabm", "numpy", "polars", "scikit-learn"]
        },
        "competition_submission": False,
    }
    (output / "results.json").write_text(json.dumps(report, indent=2) + "\n")
    print(
        f"Best validation ROC AUC: {fitted['validation_auc']:.6f}; "
        f"epoch: {fitted['best_epoch']}; training: {fitted['training_seconds']:.1f}s; "
        "no submission performed",
        flush=True,
    )

```

```python
import argparse
import gc
import hashlib
from importlib.metadata import version
import json
from pathlib import Path
import time

import joblib
import numpy as np
import polars as pl
import torch
from torch import nn
from torch.nn import functional as F



def pack_vectors(frame, fields):
    packed, missing = [], {}
    for field in fields:
        columns = [f"{field}_dim_{index}" for index in range(32)]
        raw = np.array(frame.select(columns).to_numpy(), dtype=np.float32, order="C")
        valid = np.isfinite(raw).all(axis=1)
        missing[field] = int((~valid).sum())
        raw[~valid] = 0
        norm = np.linalg.norm(raw.astype(np.float64), axis=1)
        unit = (raw / np.maximum(norm, 1e-8)[:, None]).astype(np.float32)
        log_norm = np.clip(np.log1p(norm), 0, 20).astype(np.float32)
        packed.append(
            np.concatenate([unit, log_norm[:, None], valid[:, None]], axis=1)
        )
    return torch.from_numpy(np.stack(packed, axis=1).astype(np.float32)), missing


def append_vectors(frame, numeric, fields):
    packed = torch.empty(
        (len(numeric), numeric.shape[1] + 34 * len(fields)), dtype=torch.float32
    )
    packed[:, : numeric.shape[1]] = numeric
    missing = {}
    for index, field in enumerate(fields):
        vector, counts = pack_vectors(frame, [field])
        start = numeric.shape[1] + 34 * index
        packed[:, start : start + 34] = vector[:, 0]
        missing.update(counts)
    return packed, missing


def prepare_custom_features(train, valid, min_category_count):
    dimension_columns = [name for name in train.columns if "_dim_" in name]
    fields = sorted(
        {name.rsplit("_dim_", 1)[0] for name in dimension_columns},
        key=lambda name: int(name.removeprefix("feature")),
    )
    if not fields or set(dimension_columns) != {
        f"{field}_dim_{index}" for field in fields for index in range(32)
    }:
        raise ValueError("Expected complete 32-dimensional vector fields")
    summary = [name for name in train.columns if "_dim_" not in name]
    x_train, c_train, x_valid, c_valid, state = prepare_features(
        train.select(summary), valid.select(summary)
    )
    mappings, cardinalities = [], []
    mapped_train = np.empty(c_train.shape, dtype=np.int64)
    mapped_valid = np.empty(c_valid.shape, dtype=np.int64)
    for index, cardinality in enumerate(state["cardinalities"]):
        counts = np.bincount(c_train[:, index].numpy(), minlength=cardinality)
        retained = np.flatnonzero(
            (counts >= min_category_count) & (np.arange(cardinality) != 0)
        )
        mapping = np.zeros(cardinality, dtype=np.int64)
        mapping[retained] = np.arange(1, len(retained) + 1)
        mapped_train[:, index] = mapping[c_train[:, index].numpy()]
        mapped_valid[:, index] = mapping[c_valid[:, index].numpy()]
        mappings.append(mapping)
        cardinalities.append(len(retained) + 1)
    n_dense = x_train.shape[1]
    x_train, missing_train = append_vectors(train, x_train, fields)
    x_valid, missing_valid = append_vectors(valid, x_valid, fields)
    state.update(
        {
            "original_cardinalities": state["cardinalities"],
            "cardinalities": cardinalities,
            "category_remappings": mappings,
            "min_category_count": min_category_count,
            "vector_fields": fields,
            "n_dense": n_dense,
            "missing_vectors_train": missing_train,
            "missing_vectors_valid": missing_valid,
            "validation_unknown_or_rare_category_rates": {
                name: float((mapped_valid[:, index] == 0).mean())
                for index, name in enumerate(state["categorical_columns"])
            },
        }
    )
    return (
        x_train, torch.from_numpy(mapped_train),
        x_valid, torch.from_numpy(mapped_valid), state,
    )


def transform_custom_features(frame, state):
    numeric, categorical = transform_features(frame, state)
    remapped = np.empty(categorical.shape, dtype=np.int64)
    for index, mapping in enumerate(state["category_remappings"]):
        remapped[:, index] = mapping[categorical[:, index].numpy()]
    numeric, _ = append_vectors(frame, numeric, state["vector_fields"])
    return numeric, torch.from_numpy(remapped)


class RecommendationNetwork(nn.Module):
    def __init__(
        self, n_dense, cardinalities, n_vectors, attention=False, positive_rate=0.05
    ):
        super().__init__()
        self.n_dense = n_dense
        self.n_vectors = n_vectors
        self.embeddings = nn.ModuleList(
            nn.Embedding(cardinality, 8, padding_idx=0)
            for cardinality in cardinalities
        )
        for embedding in self.embeddings:
            nn.init.normal_(embedding.weight, std=0.02)
            with torch.no_grad():
                embedding.weight[0].zero_()
        self.scalar_branch = nn.Sequential(
            nn.Linear(n_dense, 128), nn.LayerNorm(128), nn.GELU(), nn.Dropout(0.2)
        )
        self.category_branch = nn.Sequential(
            nn.Linear(8 * len(cardinalities), 64),
            nn.LayerNorm(64), nn.GELU(), nn.Dropout(0.3),
        )
        self.vector_weights = nn.Parameter(torch.empty(n_vectors, 33, 32))
        for weight in self.vector_weights:
            nn.init.xavier_uniform_(weight)
        self.vector_bias = nn.Parameter(torch.zeros(n_vectors, 32))
        self.missing_tokens = nn.Parameter(torch.randn(n_vectors, 32) * 0.02)
        self.field_tokens = nn.Parameter(torch.randn(n_vectors, 32) * 0.02)
        self.vector_transform = nn.Sequential(
            nn.LayerNorm(32), nn.Linear(32, 64), nn.GELU(),
            nn.Dropout(0.2), nn.Linear(64, 32),
        )
        self.vector_branch = nn.Sequential(
            nn.Linear(n_vectors * 32, 64),
            nn.LayerNorm(64), nn.GELU(), nn.Dropout(0.2),
        )
        self.head = nn.Sequential(
            nn.Linear(256, 64), nn.GELU(), nn.Dropout(0.3), nn.Linear(64, 1)
        )
        with torch.no_grad():
            self.head[-1].bias.fill_(float(np.log(positive_rate / (1 - positive_rate))))
        self.attention = (
            nn.MultiheadAttention(32, 4, dropout=0.2, batch_first=True)
            if attention else None
        )
        self.attention_norm = nn.LayerNorm(32) if attention else None
        self.attention_dropout = nn.Dropout(0.2)

    def extract_features(self, numeric, categorical):
        if self.training:
            categorical = categorical.masked_fill(
                torch.rand_like(categorical, dtype=torch.float32) < 0.3, 0
            )
        embedded = torch.cat(
            [
                layer(categorical[:, index])
                for index, layer in enumerate(self.embeddings)
            ],
            dim=1,
        )
        vectors = numeric[:, self.n_dense :].reshape(-1, self.n_vectors, 34)
        projected = F.gelu(
            torch.einsum("bfi,fij->bfj", vectors[:, :, :33], self.vector_weights)
            + self.vector_bias
        )
        tokens = torch.where(
            vectors[:, :, 33:].bool(), projected, self.missing_tokens.unsqueeze(0)
        ) + self.field_tokens
        if self.attention is not None:
            normalized = self.attention_norm(tokens)
            interactions, _ = self.attention(
                normalized, normalized, normalized, need_weights=False
            )
            tokens = tokens + self.attention_dropout(interactions)
        tokens = tokens + self.vector_transform(tokens)
        return torch.cat(
            [
                self.scalar_branch(numeric[:, : self.n_dense]),
                self.category_branch(embedded),
                self.vector_branch(tokens.flatten(1)),
            ],
            dim=1,
        )
    def forward(self, numeric, categorical):
        return self.head(self.extract_features(numeric, categorical))


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--rows", type=int, default=150_000)
    parser.add_argument("--data-dir", type=Path, default=ROOT / "data")
    parser.add_argument("--sample-path", type=Path)
    parser.add_argument("--epochs", type=int, default=30)
    parser.add_argument("--patience", type=int, default=5)
    parser.add_argument("--batch-size", type=int, default=1024)
    parser.add_argument("--min-category-count", type=int, default=10)
    parser.add_argument("--variant", choices=["both", "control", "attention"], default="both")
    parser.add_argument("--device", choices=["auto", "cpu", "mps", "cuda"], default="auto")
    parser.add_argument("--threads", type=int, default=4)
    parser.add_argument("--output-dir", type=Path, default=ROOT / "artifacts/local-custom-v1")
    args = parser.parse_args()
    if min(
        args.rows, args.epochs, args.patience, args.batch_size,
        args.min_category_count, args.threads,
    ) <= 0:
        parser.error("All numerical arguments must be positive")
    device = args.device
    if device == "auto":
        device = (
            "cuda" if torch.cuda.is_available()
            else "mps" if torch.backends.mps.is_available()
            else "cpu"
        )
    if device == "cuda" and not torch.cuda.is_available():
        raise RuntimeError("CUDA requested but unavailable")
    if device == "mps" and not torch.backends.mps.is_available():
        raise RuntimeError("Apple MPS requested but unavailable")
    torch.set_num_threads(args.threads)
    started = time.monotonic()
    sample_hash = None
    if args.sample_path:
        sample = pl.read_parquet(args.sample_path)
        if sample.height != args.rows:
            raise ValueError("Cached sample row count does not match --rows")
        with args.sample_path.open("rb") as stream:
            sample_hash = hashlib.file_digest(stream, "sha256").hexdigest()
    else:
        sample = load_sample(
            args.data_dir / "train.parquet", args.rows, 42, expand_embeddings=True
        )
    sample = sample.sort("_time")
    if sample.height < 2 or sample["_time"].null_count():
        raise ValueError("Chronological validation requires nonmissing timestamps")
    cutoff = sample["_time"][int(sample.height * 0.8)]
    split = int(sample["_time"].search_sorted(cutoff, side="left"))
    train = sample.slice(0, split)
    valid = sample.slice(split)
    for label, frame in [("training", train), ("validation", valid)]:
        if set(frame[TARGET].unique().to_list()) != {0, 1}:
            raise ValueError(f"{label} must contain both binary target classes")
    x_train, c_train, x_valid, c_valid, state = prepare_custom_features(
        train, valid, args.min_category_count
    )
    y_train = torch.tensor(train[TARGET].to_numpy().copy(), dtype=torch.float32)
    y_valid = valid[TARGET].to_numpy().copy()
    report = {
        "seed": 42,
        "sample_rows": sample.height,
        "sample_sha256": sample_hash,
        "training_rows": train.height,
        "validation_rows": valid.height,
        "cutoff": str(cutoff),
        "input_features": sample.width - 2,
        "dense_channels_after_preprocessing": state["n_dense"],
        "vector_fields": state["vector_fields"],
        "original_cardinalities": state["original_cardinalities"],
        "retained_cardinalities": state["cardinalities"],
        "min_category_count": args.min_category_count,
        "validation_unknown_or_rare_category_rates": state["validation_unknown_or_rare_category_rates"],
        "missing_vectors_train": state["missing_vectors_train"],
        "missing_vectors_valid": state["missing_vectors_valid"],
        "device": device,
        "threads": args.threads,
        "max_epochs": args.epochs,
        "patience": args.patience,
        "batch_size": args.batch_size,
        "optimizer": {"name": "AdamW", "learning_rate": 0.0005, "weight_decay": 0.01},
        "versions": {
            name: version(name)
            for name in ["torch", "numpy", "polars", "scikit-learn"]
        },
        "competition_submission": False,
        "results": [],
    }
    positive_rate = float(y_train.mean().item())
    del sample, train, valid
    gc.collect()
    output = args.output_dir
    output.mkdir(parents=True, exist_ok=True)
    joblib.dump(state, output / "preprocessing.joblib")
    report["preparation_seconds"] = time.monotonic() - started
    variants = ["control", "attention"] if args.variant == "both" else [args.variant]
    for variant in variants:
        torch.manual_seed(42)
        model_config = {
            "n_dense": state["n_dense"],
            "cardinalities": state["cardinalities"],
            "n_vectors": len(state["vector_fields"]),
            "attention": variant == "attention",
            "positive_rate": positive_rate,
        }
        model = RecommendationNetwork(**model_config).to(device)
        optimizer = torch.optim.AdamW(model.parameters(), lr=0.0005, weight_decay=0.01)
        variant_output = output / variant
        variant_output.mkdir(exist_ok=True)
        parameters = sum(parameter.numel() for parameter in model.parameters())
        print(
            f"Variant: {variant}; device: {device}; train: {len(y_train):,}; "
            f"valid: {len(y_valid):,}; parameters: {parameters:,}; cutoff: {cutoff}",
            flush=True,
        )
        fitted = fit_model(
            model, model_config, optimizer, x_train, c_train, y_train,
            x_valid, c_valid, y_valid, device=device, output=variant_output,
            epochs=args.epochs, patience=args.patience, batch_size=args.batch_size,
        )
        result = {
            "variant": variant,
            "parameters": parameters,
            "model_config": model_config,
            **fitted,
        }
        report["results"].append(result)
        (output / "results.json").write_text(json.dumps(report, indent=2) + "\n")
        print(
            f"{variant}: AUC={fitted['validation_auc']:.6f}; "
            f"epoch={fitted['best_epoch']}; training={fitted['training_seconds']:.1f}s",
            flush=True,
        )
        del model, optimizer
        gc.collect()
        if device == "mps":
            torch.mps.empty_cache()
        elif device == "cuda":
            torch.cuda.empty_cache()
    print("Requested validation results saved; no competition submission performed", flush=True)

```

```python
import math

import numpy as np
import torch
from torch import nn
from torch.nn import functional as F


def pairwise_vector_features(vectors, pairs):
    first, second = pairs
    unit = vectors[:, :, :32]
    cosine = (unit[:, first] * unit[:, second]).sum(dim=-1).clamp(-1, 1)
    norms = vectors[:, :, 32].expm1()
    present = vectors[:, first, 33] * vectors[:, second, 33]
    dot = cosine * norms[:, first] * norms[:, second]
    distance = (
        norms[:, first].square() + norms[:, second].square() - 2 * dot
    ).clamp_min(0).sqrt()
    return torch.cat([
        cosine * present,
        dot.sign() * dot.abs().log1p().clamp_max(20) * present,
        distance.log1p().clamp_max(20) * present,
        present,
    ], dim=1)


class FeatureEncoder(nn.Module):
    def __init__(
        self, n_numeric, n_dense, n_vectors, cardinalities, continuous_indices,
        embedding_dimensions, numerical_encoding="none", bins=None,
        vector_projection=False, vector_dropout=0.0, vector_crosses=False,
        summary_groups=None, indicator_groups=None, numeric_dropout=0.0,
    ):
        super().__init__()
        self.n_dense, self.n_vectors = n_dense, n_vectors
        self.vector_dropout, self.vector_crosses = vector_dropout, vector_crosses
        self.numeric_dropout = numeric_dropout
        self.summary_groups = summary_groups or [[] for _ in range(n_vectors)]
        self.indicator_groups = indicator_groups or [[] for _ in range(n_vectors)]
        self.register_buffer("continuous", torch.tensor(continuous_indices, dtype=torch.int64))
        remaining = [i for i in range(n_numeric) if i not in set(continuous_indices)]
        self.register_buffer("remaining", torch.tensor(remaining, dtype=torch.int64))
        self.embeddings = nn.ModuleList(
            nn.Embedding(cardinality, dimension, padding_idx=0)
            for cardinality, dimension in zip(cardinalities, embedding_dimensions)
        )
        for layer in self.embeddings:
            nn.init.normal_(layer.weight, std=0.02)
            with torch.no_grad():
                layer.weight[0].zero_()
        self.numerical = None
        dimension = n_numeric + sum(embedding_dimensions)
        if numerical_encoding != "none":
            from rtdl_num_embeddings import PeriodicEmbeddings, PiecewiseLinearEmbeddings

            if not continuous_indices:
                raise ValueError("Numerical embeddings require continuous scalar features")
            if numerical_encoding == "ple":
                if bins is None or len(bins) != len(continuous_indices):
                    raise ValueError("PLE requires one train-fitted bin array per feature")
                self.numerical = PiecewiseLinearEmbeddings(
                    [torch.tensor(values, dtype=torch.float32) for values in bins],
                    8, activation=True, version="B",
                )
            elif numerical_encoding == "plr":
                self.numerical = PeriodicEmbeddings(
                    len(continuous_indices), 8, n_frequencies=8,
                    frequency_init_scale=0.01, activation=True, lite=True,
                )
            else:
                raise ValueError(f"Unknown numerical encoding: {numerical_encoding}")
            dimension += len(continuous_indices) * 7
        self.vector_weight = None
        if vector_projection:
            self.vector_weight = nn.Parameter(torch.empty(n_vectors, 32, 32))
            for matrix in self.vector_weight:
                nn.init.xavier_uniform_(matrix)
            self.vector_bias = nn.Parameter(torch.zeros(n_vectors, 32))
            self.vector_norm = nn.LayerNorm(32)
        pairs = torch.triu_indices(n_vectors, n_vectors, offset=1)
        self.register_buffer("pairs", pairs)
        if vector_crosses:
            dimension += pairs.shape[1] * 4
        self.output_dim = dimension

    def forward(self, numeric, categorical):
        if self.training and self.numeric_dropout:
            numeric = numeric.clone()
            numeric[:, self.continuous] = F.dropout(
                numeric[:, self.continuous], p=self.numeric_dropout, training=True
            )
        vectors = numeric[:, self.n_dense :].reshape(-1, self.n_vectors, 34)
        if self.training and self.vector_dropout:
            keep = torch.rand(
                vectors.shape[:2], device=vectors.device
            ) >= self.vector_dropout
            numeric = numeric.clone()
            vectors = vectors * keep.unsqueeze(-1)
            for index in range(self.n_vectors):
                if self.summary_groups[index]:
                    numeric[:, self.summary_groups[index]] *= keep[:, index : index + 1]
                if self.indicator_groups[index]:
                    columns = self.indicator_groups[index]
                    numeric[:, columns] = torch.where(
                        keep[:, index : index + 1],
                        numeric[:, columns], torch.ones_like(numeric[:, columns]),
                    )
        crossed = None
        if self.vector_crosses:
            crossed = pairwise_vector_features(vectors, self.pairs)
        if self.vector_weight is not None:
            directions = F.silu(self.vector_norm(
                torch.einsum("bfi,fij->bfj", vectors[:, :, :32], self.vector_weight)
                + self.vector_bias
            )) * vectors[:, :, 33:]
            vectors = torch.cat([directions, vectors[:, :, 32:]], dim=-1)
        if self.vector_weight is not None or (self.training and self.vector_dropout):
            numeric = torch.cat([numeric[:, : self.n_dense], vectors.flatten(1)], dim=1)
        if self.numerical is not None:
            numeric = torch.cat([
                numeric[:, self.remaining],
                self.numerical(numeric[:, self.continuous]).flatten(1),
            ], dim=1)
        if self.training:
            categorical = categorical.masked_fill(
                torch.rand_like(categorical, dtype=torch.float32) < 0.3, 0
            )
        parts = [
            numeric,
            *[layer(categorical[:, index]) for index, layer in enumerate(self.embeddings)],
        ]
        if crossed is not None:
            parts.append(crossed)
        return torch.cat(parts, dim=1)


class LowRankCross(nn.Module):
    def __init__(self, dimension, rank=32, experts=1):
        super().__init__()
        self.experts = experts
        self.down = nn.ModuleList(nn.Linear(dimension, rank, bias=False) for _ in range(experts))
        self.up = nn.ModuleList(nn.Linear(rank, dimension, bias=False) for _ in range(experts))
        self.bias = nn.Parameter(torch.zeros(experts, dimension))
        if experts > 1:
            self.middle = nn.ModuleList(nn.Linear(rank, rank, bias=False) for _ in range(experts))
            self.gate = nn.Linear(dimension, experts, bias=False)

    def forward(self, original, current):
        if self.experts == 1:
            return current + original * (self.up[0](self.down[0](current)) + self.bias[0])
        weights = self.gate(current).softmax(dim=-1)
        transformed = []
        for index in range(self.experts):
            low = self.down[index](current).tanh()
            low = self.middle[index](low).tanh()
            transformed.append(
                original * (self.up[index](low) + self.bias[index])
            )
        return current + (
            torch.stack(transformed, dim=1) * weights.unsqueeze(-1)
        ).sum(dim=1)


class DCNv2Backbone(nn.Module):
    def __init__(self, dimension, width=128, dropout=0.2, rank=32, experts=1):
        super().__init__()
        self.crosses = nn.ModuleList(LowRankCross(dimension, rank, experts) for _ in range(2))
        self.deep = nn.Sequential(
            nn.Linear(dimension, width), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(width, width), nn.ReLU(), nn.Dropout(dropout),
        )
        self.output = nn.Linear(dimension + width, 1)

    def forward(self, inputs):
        crossed = inputs
        for layer in self.crosses:
            crossed = layer(inputs, crossed)
        return self.output(torch.cat([crossed, self.deep(inputs)], dim=1))


class FinalMLPBackbone(nn.Module):
    def __init__(self, dimension, width=128, dropout=0.2, heads=4, gating=True):
        super().__init__()
        if width % heads:
            raise ValueError("FinalMLP width must be divisible by head count")
        self.width, self.heads, self.gating = width, heads, gating
        self.streams = nn.ModuleList(
            nn.Sequential(
                nn.Linear(dimension, width), nn.ReLU(), nn.Dropout(dropout),
                nn.Linear(width, width), nn.ReLU(), nn.Dropout(dropout),
            ) for _ in range(2)
        )
        if gating:
            self.gates = nn.ModuleList(
                nn.Sequential(
                    nn.Linear(dimension, 64), nn.ReLU(),
                    nn.Linear(64, dimension), nn.Sigmoid(),
                ) for _ in range(2)
            )
        self.linear = nn.ModuleList(nn.Linear(width, 1, bias=False) for _ in range(2))
        self.bilinear = nn.Parameter(torch.empty(heads, width // heads, width // heads))
        for matrix in self.bilinear:
            nn.init.xavier_uniform_(matrix)
        self.bias = nn.Parameter(torch.zeros(1))

    def forward(self, inputs):
        streams = [
            stream(inputs * (2 * self.gates[index](inputs) if self.gating else 1))
            for index, stream in enumerate(self.streams)
        ]
        first, second = [stream.reshape(-1, self.heads, self.width // self.heads) for stream in streams]
        interaction = torch.einsum("bhd,hde,bhe->b", first, self.bilinear, second).unsqueeze(1)
        return self.linear[0](streams[0]) + self.linear[1](streams[1]) + interaction + self.bias


class HeterogeneousClassifier(nn.Module):
    def __init__(
        self, encoder_config, architecture="tabm", width=128, dropout=0.2,
        members=8, positive_rate=0.05, rank=32, experts=1, gating=True,
    ):
        super().__init__()
        self.encoder = FeatureEncoder(**encoder_config)
        dimension = self.encoder.output_dim
        if architecture == "tabm":
            from tabm import TabM

            self.backbone = TabM.make(
                n_num_features=dimension, d_out=1, k=members,
                n_blocks=2, d_block=width, dropout=dropout,
            )
            bias = self.backbone.output.bias
        elif architecture == "dcnv2":
            self.backbone = DCNv2Backbone(dimension, width, dropout, rank, experts)
            bias = self.backbone.output.bias
        elif architecture == "finalmlp":
            self.backbone = FinalMLPBackbone(dimension, width, dropout, gating=gating)
            bias = self.backbone.bias
        else:
            raise ValueError(f"Unknown architecture: {architecture}")
        with torch.no_grad():
            bias.fill_(float(np.log(positive_rate / (1 - positive_rate))))
        self.architecture = architecture

    def forward(self, numeric, categorical):
        output = self.backbone(self.encoder(numeric, categorical))
        return output.squeeze(-1) if self.architecture == "tabm" else output
```

```python
"""Label-free within-request context features appended to raw ShareChat parquet files.

Every row of a request shares one timestamp and requests never cross the
train/holdout/test boundaries, so these features are computed per request and
use no labels. request_id itself is never a model input.
"""
import argparse
from pathlib import Path

import numpy as np
import polars as pl
import pyarrow as pa
import pyarrow.parquet as pq


ITEM = "feature18"
RANKED = ["feature418", "feature419", "feature411", "feature410", "feature430",
          "feature431", "feature417", "feature352", "feature346", "feature480"]


def request_expressions():
    group = "request_id"
    size = pl.len().over(group)
    exprs = [size.alias("rq_size"),
             pl.col(ITEM).n_unique().over(group).alias("rq_items"),
             pl.len().over([group, ITEM]).alias("rq_item_dup")]
    for c in RANKED:
        exprs += [
            (pl.col(c).rank("average").over(group) / size).alias(f"rq_rank_{c}"),
            (pl.col(c) - pl.col(c).mean().over(group)).alias(f"rq_dev_{c}"),
            (pl.col(c) - pl.col(c).max().over(group)).alias(f"rq_gap_{c}"),
        ]
    return [e.cast(pl.Float32) for e in exprs]


def request_features(frame):
    """Return request features in the same row order as ``frame``."""
    return frame.select(request_expressions())


def augment(source, destination, batch_size=65536):
    """Copy ``source`` to ``destination`` with request features appended, preserving row order."""
    narrow = pl.read_parquet(source, columns=["request_id", ITEM, *RANKED])
    extra = request_features(narrow)
    del narrow
    parquet = pq.ParquetFile(source)
    if parquet.metadata.num_rows != len(extra):
        raise ValueError("Request features do not cover every source row")
    overlap = set(extra.columns) & set(parquet.schema_arrow.names)
    if overlap:
        raise ValueError(f"Source already contains request features: {sorted(overlap)}")
    destination.parent.mkdir(parents=True, exist_ok=True)
    writer, offset = None, 0
    try:
        for batch in parquet.iter_batches(batch_size=batch_size):
            table = pa.Table.from_batches([batch])
            appended = extra.slice(offset, len(table)).to_arrow()
            for name in appended.column_names:
                table = table.append_column(name, appended[name])
            if writer is None:
                writer = pq.ParquetWriter(destination, table.schema)
            writer.write_table(table)
            offset += len(table)
    finally:
        if writer is not None:
            writer.close()
    if offset != len(extra):
        raise ValueError("Augmented parquet does not cover every source row")
    return extra.columns


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--data-dir", type=Path, required=True)
    parser.add_argument("--output-dir", type=Path, required=True)
    args = parser.parse_args()
    for name in ["train", "test"]:
        columns = augment(args.data_dir / f"{name}.parquet", args.output_dir / f"{name}.parquet")
        print(f"{name}: appended {len(columns)} request features", flush=True)
    sample = args.data_dir / "sample_submission.csv"
    target = args.output_dir / "sample_submission.csv"
    if sample.is_file() and not target.exists():
        target.symlink_to(sample.resolve())

```

```python
import argparse
from datetime import datetime
import hashlib
from importlib.metadata import version
import json
from pathlib import Path
import subprocess
import sys
import tempfile
import time

import joblib
import numpy as np
import polars as pl
import pyarrow.parquet as pq
from sklearn.metrics import roc_auc_score
import torch



def digest(path):
    with path.open("rb") as stream:
        return hashlib.file_digest(stream, "sha256").hexdigest()


def combine(probabilities, weights):
    if set(probabilities) != set(weights) or abs(sum(weights.values()) - 1) > 1e-12:
        raise ValueError("Ensemble membership or weights differ")
    return sum(weight * probabilities[name].astype(np.float64) for name, weight in weights.items())


def run(bundle, data, output, *, device="cuda", batch_size=1024, worker_path=None):
    started = time.monotonic()
    manifest = json.loads((bundle / "manifest.json").read_text())
    for name, expected in manifest["files_sha256"].items():
        if digest(bundle / name) != expected:
            raise ValueError(f"Model bundle checksum mismatch: {name}")
    versions = {name: version(name) for name in manifest["versions"]}
    if versions != manifest["versions"]:
        raise RuntimeError(f"Use the original training environment: expected {manifest['versions']}; got {versions}")
    torch.set_num_threads(4)
    state = joblib.load(bundle / "preprocessing.joblib")
    state = {**state, "category_remappings": joblib.load(bundle / "category-remappings.joblib")}
    models = {}
    for name, recorded in manifest["neural_models"].items():
        checkpoint = torch.load(bundle / recorded["checkpoint"], map_location="cpu", weights_only=True)
        if checkpoint["model_config"] != recorded["model_config"]:
            raise ValueError(f"Checkpoint configuration mismatch: {name}")
        model = HeterogeneousClassifier(**checkpoint["model_config"]).to(device)
        model.load_state_dict(checkpoint["state_dict"], strict=True)
        model.eval()
        models[name] = model
    columns = manifest["tree_columns"]
    vocabularies = json.loads((bundle / "category-vocabularies.json").read_text())
    category_expressions = [
        pl.col(name).replace_strict(
            {value: index for index, value in enumerate(values)}, default=-1, return_dtype=pl.Int32,
        )
        for name, values in vocabularies.items()
    ]
    output.mkdir(parents=True, exist_ok=True)
    cutoff = datetime.fromisoformat(manifest["cutoff"])
    template = pl.read_csv(data / "sample_submission.csv")
    if template.columns != ["id", TARGET] or template["id"].n_unique() != len(template):
        raise ValueError("Invalid sample-submission template")
    if pq.ParquetFile(data / "test.parquet").metadata.num_rows != len(template):
        raise ValueError("Test and sample-submission row counts differ")

    def infer(path, count, *, validation):
        predictions = {name: [] for name in models}
        identities, labels, offset, written = [], [], 0, 0
        with tempfile.TemporaryDirectory(prefix="finalist-inference-", dir=output) as directory:
            temporary = Path(directory)
            matrix = np.lib.format.open_memmap(
                temporary / "features.npy", mode="w+", dtype=np.float32,
                shape=(count, len(columns)),
            )
            for batch in pq.ParquetFile(path).iter_batches(batch_size=8192):
                raw = pl.from_arrow(batch).with_columns(
                    pl.Series("_source_row_id", np.arange(offset, offset + batch.num_rows)),
                )
                offset += len(raw)
                if validation:
                    raw = raw.filter(pl.col("feature1") >= cutoff)
                    if raw.is_empty():
                        continue
                    identities.append(raw["_source_row_id"].to_numpy())
                    labels.append(raw[TARGET].to_numpy())
                else:
                    if raw["id"].to_list() != template["id"].slice(written, len(raw)).to_list():
                        raise ValueError("Test IDs differ from the submission template")
                    identities.append(raw["id"].to_numpy())
                frame = features(raw.drop("_source_row_id"), expand_embeddings=True)
                if set(frame.columns) != set(columns) or len(frame.columns) != len(columns):
                    raise ValueError("Raw features differ from the training feature schema")
                frame = frame.select(columns)
                numeric, categorical = transform_custom_features(frame, state)
                for name, model in models.items():
                    predictions[name].append(predict(model, numeric, categorical, device, batch_size))
                encoded = frame.with_columns(category_expressions).select(
                    pl.all().cast(pl.Float32)
                ).to_numpy()
                if written + len(raw) > count:
                    raise ValueError("Prediction population exceeds the expected count")
                matrix[written:written + len(raw)] = encoded
                written += len(raw)
                if written % (8192 * 20) == 0:
                    print(f"{'Validation' if validation else 'Test'}: {written:,}/{count:,}", flush=True)
            if written != count:
                raise ValueError(f"Prediction population mismatch: {written} != {count}")
            matrix.flush()
            del matrix
            (temporary / "columns.json").write_text(json.dumps(columns))
            subprocess.run([
                sys.executable, str(worker_path or Path(__file__).with_name("lightgbm_worker.py")),
                "--predict-array", str(temporary / "features.npy"),
                "--model-file", str(bundle / "lightgbm.txt"),
                "--feature-columns", str(temporary / "columns.json"),
                "--output-dir", str(temporary), "--threads", "2",
            ], check=True)
            probabilities = {name: np.concatenate(parts) for name, parts in predictions.items()}
            probabilities["lightgbm"] = np.load(temporary / "probabilities.npy").copy()
        for name, values in probabilities.items():
            if len(values) != count or not np.isfinite(values).all() or not ((values >= 0) & (values <= 1)).all():
                raise ValueError(f"Invalid probabilities: {name}")
        return probabilities, np.concatenate(identities), np.concatenate(labels) if validation else None

    validation, row_ids, labels = infer(
        data / "train.parquet", manifest["validation_rows"], validation=True,
    )
    with np.load(bundle / "validation-reference.npz") as reference:
        sorted_indices = np.argsort(row_ids)
        np.testing.assert_array_equal(np.sort(row_ids), np.sort(reference["row_ids"]))
        order = sorted_indices[np.searchsorted(row_ids[sorted_indices], reference["row_ids"])]
        np.testing.assert_array_equal(row_ids[order], reference["row_ids"])
        np.testing.assert_array_equal(labels[order], reference["y_true"])
        validation = {name: values[order] for name, values in validation.items()}
        validation["ensemble"] = combine(validation, manifest["ensemble_weights"])
        replay = {}
        for name, values in validation.items():
            expected = reference[name]
            error = float(np.max(np.abs(values - expected)))
            auc = float(roc_auc_score(reference["y_true"], values))
            expected_auc = float(roc_auc_score(reference["y_true"], expected))
            if error > 0.00005 or abs(auc - expected_auc) > 0.000005:
                raise ValueError(f"Validation replay mismatch for {name}: max error {error}, AUC {auc} != {expected_auc}")
            replay[name] = {"validation_auc": auc, "reference_auc": expected_auc, "max_probability_error": error}
        np.savez_compressed(
            output / "validation-replay.npz", y_true=reference["y_true"],
            row_ids=reference["row_ids"], **validation,
        )
    print(f"Full validation replay passed: {json.dumps(replay)}", flush=True)
    test, test_ids, _ = infer(data / "test.parquet", len(template), validation=False)
    np.testing.assert_array_equal(test_ids, template["id"].to_numpy())
    test["ensemble"] = combine(test, manifest["ensemble_weights"])
    submissions = {}
    for name, values in [("tabm", test["tabm32"]), ("ensemble", test["ensemble"])]:
        destination = output / f"submission-{name}.csv"
        template.with_columns(pl.Series(TARGET, values)).write_csv(destination)
        submissions[name] = {
            "file": destination.name, "sha256": digest(destination),
            "rows": len(values), "minimum": float(values.min()),
            "mean": float(values.mean()), "maximum": float(values.max()),
        }
    report = {
        "status": "COMPLETE", "refit": False, "automatic_competition_submission": False,
        "bundle_manifest_sha256": digest(bundle / "manifest.json"),
        "versions": versions, "validation_replay": replay, "submissions": submissions,
        "ensemble_weights": manifest["ensemble_weights"],
        "inference_seconds": time.monotonic() - started,
    }
    (output / "inference-results.json").write_text(json.dumps(report, indent=2) + "\n")
    print(json.dumps(report, indent=2), flush=True)
    return report


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--bundle", type=Path, required=True)
    parser.add_argument("--data-dir", type=Path, required=True)
    parser.add_argument("--output-dir", type=Path, required=True)
    parser.add_argument("--device", choices=["cpu", "cuda"], default="cuda")
    parser.add_argument("--batch-size", type=int, default=1024)
    args = parser.parse_args()
    if args.batch_size <= 0 or (args.device == "cuda" and not torch.cuda.is_available()):
        parser.error("A positive batch size and the requested CUDA device are required")
    run(
        args.bundle, args.data_dir, args.output_dir, device=args.device,
        batch_size=args.batch_size, worker_path=globals().get("FROZEN_LIGHTGBM_WORKER_FILE"),
    )

```
